# Market Expansion Toolkit — Starter Analysis
This notebook loads a city scorecard, calculates a simple priority ranking, and visualizes the results.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../scorecards/jakarta_scorecard.csv')
df
```

```python
df['Calculated_Score'] = (
    df['Demand'] + df['Supply'] + df['Ops_Feasibility']
    - df['Competition'] - df['Regulatory_Friction']
)
df.sort_values('Calculated_Score', ascending=False)
```

```python
ranked = df.sort_values('Calculated_Score', ascending=True)
plt.figure(figsize=(9, 5))
plt.barh(ranked['Area'], ranked['Calculated_Score'])
plt.xlabel('Priority Score')
plt.ylabel('Area')
plt.title('Jakarta Market Expansion Priority')
plt.tight_layout()
plt.show()
```
